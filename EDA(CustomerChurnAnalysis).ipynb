```python
cols = [
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

# Create a 3x3 grid of subplots
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(15, 10))
axes = axes.flatten()  # Flatten the 2D array into 1D for easy looping

for i, col in enumerate(cols):
    ax = sns.countplot(x=col, data=df,hue='Churn', ax=axes[i])
    ax.set_title(f'Churn by {col}', fontsize=12, fontweight='bold')
    ax.set_xlabel('')
    ax.set_ylabel('Count')

    # Add value labels on top of both blue and orange bars
    for container in ax.containers:
        ax.bar_label(container, fontsize=9)
    ax.margins(y=0.12)

plt.tight_layout()
plt.show()
```

```python
# Customers with fiber optic internet churn at a much higher rate than DSL users, pointing to potential pricing or service satisfaction issues.
# Subscribers without internet service are the most loyal segment and rarely leave.
# Customers who lack Online Security or Tech Support churn at high rates, whereas those who add these services stay much longer.
# Bundling security and tech support add-ons into fiber optic plans is the most effective service-level strategy to reduce churn.
```
